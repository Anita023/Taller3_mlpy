## Pandas
Es una biblioteca de código abierto para Python que actúa como una potente hoja de cálculo virtual, diseñada específicamente para el análisis y la manipulación de datos. Su herramienta principal es el DataFrame, una estructura en forma de tabla con filas y columnas que permite importar fácilmente archivos externos (como Excel, CSV o bases de datos SQL) para procesar millones de registros en segundos. Los programadores y científicos de datos la utilizan como su herramienta principal para limpiar datos desordenados, filtrar información específica, rellenar valores faltantes y agrupar estadísticas, transformando datos crudos en información estructurada y lista para tomar decisiones o crear gráficos.

```python
import pandas as pd
import requests
from io import StringIO

url = "https://es.wikipedia.org/wiki/Anexo:Pa%C3%ADses_y_territorios_dependientes_por_poblaci%C3%B3n"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}

response = requests.get(url, headers=headers, timeout=30)
response.raise_for_status()

# Leer tablas pasando el contenido HTML con StringIO
tables = pd.read_html(StringIO(response.text))
df = tables[0]

print(df.head())
```
