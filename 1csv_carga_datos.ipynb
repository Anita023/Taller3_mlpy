## Pandas
Es una biblioteca de código abierto para Python que actúa como una potente hoja de cálculo virtual, diseñada específicamente para el análisis y la manipulación de datos. Su herramienta principal es el DataFrame, una estructura en forma de tabla con filas y columnas que permite importar fácilmente archivos externos (como Excel, CSV o bases de datos SQL) para procesar millones de registros en segundos. Los programadores y científicos de datos la utilizan como su herramienta principal para limpiar datos desordenados, filtrar información específica, rellenar valores faltantes y agrupar estadísticas, transformando datos crudos en información estructurada y lista para tomar decisiones o crear gráficos.

```python
# ==========================================
# 1. IMPORTAR LIBRERÍAS
# ==========================================

import pandas as pd
import numpy as np 


# ==========================================
# 2. CARGAR EL ARCHIVO CSV
# ==========================================

df = pd.read_csv("dataset_ventas.csv")


# ==========================================
# 3. MOSTRAR LAS PRIMERAS 5 FILAS
# ==========================================

print("Primeras 5 filas:")
display(df.head())


# ==========================================
# 4. MOSTRAR LAS ÚLTIMAS 5 FILAS
# ==========================================

print("Últimas 5 filas:")
display(df.tail())


# ==========================================
# 5. MOSTRAR 5 FILAS ALEATORIAS
# ==========================================

print("5 filas aleatorias:")
display(df.sample(n=5))


# ==========================================
# 6. INFORMACIÓN GENERAL DEL DATAFRAME
# ==========================================

print("Información general:")
df.info()


# ==========================================
# 7. ESTADÍSTICAS DESCRIPTIVAS
# ==========================================

print("Estadísticas descriptivas:")
display(df.describe())


# ==========================================
# 8. CANTIDAD DE FILAS Y COLUMNAS
# ==========================================

print("Cantidad de filas y columnas:")
print(df.shape)


# ==========================================
# 9. NOMBRES DE LAS COLUMNAS
# ==========================================

print("Nombres de las columnas:")
print(df.columns.tolist())


# ==========================================
# 10. VALORES NULOS
# ==========================================

print("Valores nulos por columna:")
print(df.isnull().sum())


# ==========================================
# 11. TIPOS DE DATOS
# ==========================================

print("Tipos de datos:")
print(df.dtypes)
```
